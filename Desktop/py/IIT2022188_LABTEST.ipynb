```python
import pandas as pd 
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score


df = pd.read_csv("diabetes.csv")
print(df.head())
print(df.info())
cols=["Glucose","BloodPressure","SkinThickness","Insulin","BMI"]
df[cols]=df[cols].replace(0,np.nan)
df.fillna(df.median(),inplace= True)
X =df.drop(columns="Outcome")
y = df["Outcome"]
sc= StandardScaler()
X_scaled =sc.fit_transform(X)
X_train,X_test,y_train,y_test=train_test_split(X_scaled,y,test_size=0.2)
kernels=['linear','rbf']
for kernel in kernels:
    SVM_model=SVC(kernel=kernel,random_state=42)
    SVM_model.fit(X_train,y_train)
    y_pred=SVM_model.predict(X_test)

accuracy= accuracy_score(y_test,y_pred)
precision= precision_score(y_test,y_pred)
F1score=f1_score(y_test,y_pred)
conf_mat= confusion_matrix(y_test,y_pred)
print("\nmetrics for svm are")
print("\naccuracy=",accuracy)
print("\nprecision",precision)
print("\nf1score",F1score)
print("\nconfusion matrix",conf_mat)



CART_tree = DecisionTreeClassifier(criterion = "gini", random_state=42)
CART_tree.fit(X_train,y_train)
y_predicted=CART_tree.predict(X_test)

accuracydt= accuracy_score(y_test,y_predicted)
precisiondt= precision_score(y_test,y_predicted)
F1scoredt=f1_score(y_test,y_predicted)
confdt= confusion_matrix(y_test,y_predicted)
print("\naccuracy for dt =",accuracydt)
print("\nprecision for dt",precisiondt)
print("\nf1score for dt",F1scoredt)
print("\nconfusionm matrix for dt",confdt)







```
